<h1>Z axis Baseline</h1>
First we examine the Z baseline data by plotting every piece of data collected when there was 0 labware on the Z axis. 
We then create a "baseline envelope" which will represent the minimum and maximum value we might expect to see for each bin. 
The envelope is calculated by finding the mean of all the values collected in one bin and then adding/subtracting 6 times the standard deviation of the values of that bin. 

We create one envelope for the baseline when the platform is in the home position and another for when it is in the gripper position.
